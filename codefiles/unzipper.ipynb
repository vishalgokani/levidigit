```python
import os

# Define paths
dataset_name = 'Dataset501_Glioblastoma'
input_folder = os.path.join(os.environ['nnUNet_raw'], dataset_name, 'imagesTs')
output_folder = os.path.join(homePath, 'inference_results')

# Check if input folder exists
if os.path.exists(input_folder):
    print(f"Found inference images in: {input_folder}")

    # Create output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Construct and run the command
    # -d 501: Dataset ID
    # -c 3d_fullres: Configuration
    # -i: Input folder
    # -o: Output folder
    # -f 0 1 2 3 4: Use all 5 folds for inference (ensemble)
    cmd = f"nnUNetv2_predict -d 501 -c 3d_fullres -i \"{input_folder}\" -o \"{output_folder}\" -f 0 1 2 3 4"
    print(f"Running command: {cmd}")
    os.system(cmd)
else:
    print(f"Error: Inference folder not found at {input_folder}. Please ensure you have an 'imagesTs' folder in your raw dataset directory.")
```
